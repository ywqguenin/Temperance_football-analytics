```python
shots = events[events["type"] == "Shot"]

shots[[
    "minute",
    "team",
    "player",
    "location",
    "shot_outcome",
    "shot_statsbomb_xg"
]]
```

```python
shots = shots.copy()

shots["x"] = shots["location"].apply(lambda x: x[0] if isinstance(x, list) else None)
shots["y"] = shots["location"].apply(lambda x: x[1] if isinstance(x, list) else None)

shots[[
    "team",
    "player",
    "x",
    "y",
    "shot_statsbomb_xg"
]].head()
```

```python
from mplsoccer import Pitch
import matplotlib.pyplot as plt

pitch = Pitch(
    pitch_type="statsbomb",
    pitch_color="white",
    line_color="black"
)

fig, ax = pitch.draw(figsize=(12, 8))

for _, shot in shots.iterrows():
    x = shot["x"]
    y = shot["y"]
    xg = shot["shot_statsbomb_xg"]

    if shot["team"] == "Arsenal":
        marker = "o"
    else:
        marker = "s"

    if shot["shot_outcome"] == "Goal":
        edgecolor = "black"
        linewidth = 2
    else:
        edgecolor = "gray"
        linewidth = 0.8

    ax.scatter(
        x,
        y,
        s=xg * 1000,
        marker=marker,
        alpha=0.6,
        edgecolors=edgecolor,
        linewidths=linewidth
    )

plt.title("Arsenal vs Manchester City — Shot Map")
plt.show()
```

```python
summary = shots.groupby("team")["shot_statsbomb_xg"].agg(
    shots="count",
    total_xg="sum",
    avg_xg="mean"
)

summary
```
